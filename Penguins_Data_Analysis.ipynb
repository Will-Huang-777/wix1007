```python
# Q1
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)

df.head()
```

```python
# Q1
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(data=df, x="species", ax=axes[0])
axes[0].set_title("Number of Penguins by Species")
axes[0].set_xlabel("Species")
axes[0].set_ylabel("Count")

sns.histplot(data=df, x="body_mass_g", bins=20, ax=axes[1])
axes[1].set_title("Body Mass Distribution")
axes[1].set_xlabel("Body Mass (g)")

sns.scatterplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species", ax=axes[2])
axes[2].set_title("Bill Length vs. Bill Depth")

plt.tight_layout()
plt.show()
```

```python
# Q6 and Q8
missing = df.isnull().sum()
missing = missing[missing > 0]

print("Number of columns with null values:", len(missing))
print("Missing values by column:")
print(missing)
print()

for col in missing.index:
    rows = df.index[df[col].isnull()].tolist()
    print(col, "->", len(rows), "missing rows")
    print("Row indices:", rows)
```
